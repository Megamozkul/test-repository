```python
n=len(df)
n1=(df['Sex']=='female').sum() ## число выборки женщин
n2=(df['Pclass']==1).sum() ## число выборки 1-ого класса
n3=(df['Sex']=='male').sum()## число выборки мужчин
```

```python
def interval(p,n):## p - вероятность выживания(т.к распределение Бернулли то оно же и будет матожиданием p=M)
    q=1-p ## q - вероятность невыживания
    sigma=(p*(1-p))**0.5 ## sigma - стандартное отклонение в выборке
    sd =sigma/(n)**0.5## sd - стандартная ошибка
    return st.norm.interval(0.95,p,sd)
res1 = interval(pa,n)
res2  = interval(fem,n1)
res3  = interval(sur1,n2)
print(res1)
print(res2)
print(res3)
```

```python
from scipy.stats import norm ## p - доверительный интервал через функцию указанную в лекции на примере общего чила паасажиров
alpha = 0.05
q=1-pa
sigma=(pa*q)**0.5
sd =sigma/(n)**0.5
def get_confidence_interval(pa, sigma, n, alpha):
    h = norm.ppf(1 - alpha / 2) * sigma / np.sqrt(n)
    return pa - h, pa + h

confidence_interval = get_confidence_interval(pa, sigma, n, alpha)

confidence_interval
```

```python
## Исходные данные
##X1= fem  вероятность выживания женщин.
##X2= man  вероятность выживания мужчин.
from scipy.stats import t
def t_test(X1,X2,n1,n2):
    q1=1-X1## вероятность невыживания женщин.
    q2=1-X2## вероятность невыживания мужчин.
    sigma1=(X1*q1)**0.5
    sigma2=(X2*q2)**0.5
    se=(((sigma1**2)/n1)+((sigma2**2)/n2))**0.5
    ## Допускаем Гипотзу H0 - в генеральной совокупности средние М1 =М2
    t_score= (X1-X2)/(se)##t -критерий 
    df=n1+n3-2 ##df -число степеней свободы
    p_value = 2 * (1 - t.cdf(np.abs(t_score), df))
    if p_value < alpha:
        return( "Можно отвергнуть нулевую гипотезу")
    else:
        return("Нулевую гипотезу отвергнуть нельзя")
T1 = t_test(fem,man,n1,n3)
T1
```
