```python
### Уровень 1 
import pandas as pd
df = pd.read_csv('C:\\Users\\User\\Documents\\ДЗ\\train.csv') 
df.head(10)
```

```python
df[df['Sex'] == 'male']['Survived'].mean()*100## процент выживших мужчин.
```

```python
df['Age'].fillna(df.groupby(['Pclass', 'Sex'])['Age'].transform('median'))
```

```python
pa=(df['Survived']==1).mean()## pa - вероятность выживания
pa
```

```python
pa2=(df['Survived']==0).mean()## pa2 - вероятность HEвыживания
pa2
```

```python
pab=df[df['Sex'] == 'female']['Survived'].mean()## pab - вероятность выживания  женщины
pab
```

```python
sur1=df[df['Pclass'] == 1]['Survived'].mean() ## вероятноость выживания 1-ого класса
sur1
```

```python
sur2=df[df['Pclass'] == 2]['Survived'].mean() ## вероятноость выживания  2-ого класса
```

```python
sur3=df[df['Pclass'] == 3]['Survived'].mean() ## пвероятноость выживания  3-ого класса
sur3
```

```python
man=df[df['Sex'] == 'male']['Survived'].mean() ## процент выживших мужчин.
man
```

```python
fem=df[df['Sex'] == 'female']['Survived'].mean() ## процент выживших женщин
fem
```

```python
pb1a=df[df['Age'] <=25]['Survived'].mean() ## pb1a - вероятность выживания с возрастом до 25 лет
print(pb1a)
pb2a=df[(df['Age'] <= 50) & (df['Age'] >25)]['Survived'].mean()## pb2a - вероятность выживания с возрастом от 25 до 50 лет
print(pb2a)
pb3a=df[df['Age'] >50]['Survived'].mean()## pb2a - вероятность выживания с возрастом боллее 50 лет
print(pb3a)
```

```python
df['выживание по полу']=df["Sex"].apply(lambda x: fem if x =='female' else man)
df['выживание по классу']=df["Pclass"].apply(lambda x: sur1 if x ==1 else( sur2 if x ==2 else sur3))
df['выживание по возрасту']=df["Age"].apply(lambda x: pb1a if x<=25 else( pb3a if x>50 else pb2a))
df
```

```python
df.drop(['Cabin','Embarked','Ticket','Parch','Fare'],axis=1,inplace=True)
```

```python
df
```

```python
pb1a2=(df[df['Age'] <=25]['Survived']==0).mean() ## pb1a - вероятность выживания с возрастом до 25 лет
print(pb1a2)
pb2a2=(df[(df['Age'] <= 50) & (df['Age'] >25)]['Survived']==0).mean()## pb2a - вероятность выживания с возрастом от 25 до 50 лет
print(pb2a2)
pb3a2=(df[df['Age'] >50]['Survived']==0).mean()## pb2a - вероятность выживания с возрастом боллее 50 лет
print(pb3a2)
```

```python
df['Вероятность выживания']=df['выживание по полу']*df['выживание по классу']*df['выживание по возрасту']*pa
df
```

```python
df['Вероятность НЕвыживания']=(1-df['выживание по полу'])*(1-df['выживание по классу'])*(1-df['выживание по возрасту'])*pa2
df['delta']=df['Вероятность выживания']-df['Вероятность НЕвыживания']
df['prognoz']=df["delta"].apply(lambda x: 0 if x <0 else 1)
df
```
