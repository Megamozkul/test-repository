```python
import numpy as np
def f(x, y):
    return (x -3)**2 + (y +5)**2 
def grad_f(x, y):
    grad_x = 2*(x-3)
    grad_y = 2 *(y+5)
    return np.array([grad_x, grad_y])
x0 = np.array([5, 2])
x_old = x0
# Эти параметры можно менять и смотреть на результат
eps = 10e-6
learning_rate = 0.1
max_steps = 100
i = 0
while np.linalg.norm(grad_f(x_old[0], x_old[1])) > eps:
# Обновляем x
    x_new = x_old - learning_rate * grad_f(x_old[0], x_old[1])
    x_old = x_new
# Увеличиваем счетчик
    i += 1
# Проверяем, нужно ли выходить из цикла
    if i == max_steps:
        break
    print(x_new, f(x_new[0], x_new[1]), i)
```

```python
import numpy as np
from scipy.optimize import minimize

#  функция
def f(v):
    x, y = v
    return (x - 3)*2 + (y + 5)*2

# градиент
def grad(v):
    x, y = v
    return np.array([2*(x - 3), 2*(y + 5)])

# начальная точка
x0 = np.array([5.0, 2.0])

path = []

# callback вызывается на каждой итерации (передаётся текущая точка)
def cb(xk):
    path.append(np.copy(xk))

res = minimize(f, x0, jac=grad, method='BFGS',
               callback=cb,
               options={'gtol': 1e-6, 'disp': False, 'maxiter': 100})

print("x =", res.x)
print("f(x) =", res.fun)

print("\n Итерации:")
for i, p in enumerate(path):
    print(f"{i+1}: {p}")
```
