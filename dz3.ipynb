```python
import numpy as np

A = np.array([[5,10,1],[0,1,0],[9,3,6]])
B = np.array([[1,7,3],[0,2,0],[9,4,2]])
D = np.array([[5,1,8],[2,6,3],[1,3,0]])

C = A @ B
E = A @ D
DT = D.T
print(C)
print()
print(E)
print()
print(DT)
```

```python
##Уровень 2
from scipy import linalg
import numpy as np

A = np.array([[100,200,1000],[350,150,55],[30,30,30]])
B = np.array([2600,1660,210])

C = linalg.solve(A,B)
print(C)
```
